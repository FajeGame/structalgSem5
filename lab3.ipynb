Возьмите реализацию класса UnorderedList из лекционных материалов и выполните
следующие доработки:
1. Реализуйте оставшиеся методы этого класса append(item), index(item),
insert(pos,item), pop(), pop(pos). Требования к работе этих методов посмотрите в
лекционной презентации.
2. Реализуйте метод __str__, чтобы содержимое списка отображалось в Python-стиле (в
квадратных скобках).
3. Реализуйте метод “среза”. Он должен принимать два параметра (start и stop) и
возвращать копию списка, начиная с позиции start и заканчивая (но не включая!)
позицией stop.
Все выполненные доработки должны быть протестированы.


```python
from typing import Optional, Any, Iterable

class Node:
    def __init__(self, data: Any):
        self.data = data
        self.next: Optional['Node'] = None

class UnorderedList:
    def __init__(self, items: Optional[Iterable[Any]] = None):
        self.head: Optional[Node] = None
        self._size = 0
        if items:
            for x in items:
                self.append(x)

    def is_empty(self) -> bool:
        return self.head is None

    def add(self, item: Any):
        n = Node(item)
        n.next = self.head
        self.head = n
        self._size += 1

    def append(self, item: Any):
        n = Node(item)
        if self.head is None:
            self.head = n
        else:
            cur = self.head
            while cur.next:
                cur = cur.next
            cur.next = n
        self._size += 1

    def size(self) -> int:
        return self._size

    def search(self, item: Any) -> bool:
        cur = self.head
        while cur:
            if cur.data == item:
                return True
            cur = cur.next
        return False

    def remove(self, item: Any):
        prev = None
        cur = self.head
        while cur:
            if cur.data == item:
                if prev is None:
                    self.head = cur.next
                else:
                    prev.next = cur.next
                self._size -= 1
                return
            prev = cur
            cur = cur.next
        raise ValueError("Item not found")

    def index(self, item: Any) -> int:
        idx = 0
        cur = self.head
        while cur:
            if cur.data == item:
                return idx
            cur = cur.next
            idx += 1
        raise ValueError("Item not found")

    def insert(self, pos: int, item: Any):
        if pos <= 0:
            self.add(item)
            return
        if pos >= self._size:
            self.append(item)
            return
        prev = None
        cur = self.head
        i = 0
        while i < pos:
            prev = cur
            cur = cur.next
            i += 1
        node = Node(item)
        prev.next = node
        node.next = cur
        self._size += 1

    def pop(self, pos: Optional[int] = None) -> Any:
        if self.is_empty():
            raise IndexError("pop from empty list")
        if pos is None:
            pos = self._size - 1
        if pos < 0 or pos >= self._size:
            raise IndexError("pop index out of range")
        prev = None
        cur = self.head
        i = 0
        while i < pos:
            prev = cur
            cur = cur.next
            i += 1
        if prev is None:
            self.head = cur.next
        else:
            prev.next = cur.next
        self._size -= 1
        return cur.data

    def slice(self, start: int, stop: int) -> 'UnorderedList':
        if start < 0:
            start = 0
        res = UnorderedList()
        cur = self.head
        i = 0
        while cur and i < start:
            cur = cur.next
            i += 1
        while cur and i < stop:
            res.append(cur.data)
            cur = cur.next
            i += 1
        return res

    def __iter__(self):
        cur = self.head
        while cur:
            yield cur.data
            cur = cur.next

    def __str__(self) -> str:
        return "[" + ", ".join(repr(x) for x in self) + "]"


ul = UnorderedList([1,2,3])
print("Изначально:", str(ul))          # [1, 2, 3]
ul.add(0)
print("После add(0):", str(ul))        # [0,1,2,3]
ul.append(4)
print("После append(4):", str(ul))     # [0,1,2,3,4]
print("index(3):", ul.index(3))        # 3
ul.insert(2, 99)
print("После insert(2,99):", str(ul))  # [0,1,99,2,3,4]
print("pop():", ul.pop(), str(ul))     # 4
print("pop(2):", ul.pop(2), str(ul))   # 99
print("slice(1,3):", str(ul.slice(1,3)))  # [1,2]
```

Напишите функцию move_to_front для чтения строк с клавиатуры и сохранения их в
связном списке без повторов. Пользователь последовательно вводит с клавиатуры n строк.
При вводе строки, которая раньше не встречалась, вставьте ее в начало списка. При вводе
уже внесенной строки удалите ее из списка и снова вставьте в начало.
Такая функция реализует стратегию «сдвиг в начало», которая применяется в кешировании,
при сжатии данных и т.д.


```python
def move_to_front_from_iter(lines):
    lst = UnorderedList()
    for s in lines:
        if lst.search(s):
            lst.remove(s)
        lst.add(s)
    return lst

input_lines = ["apple", "banana", "apple", "cherry", "banana"]
res_lst = move_to_front_from_iter(input_lines)
print("Ввод:", input_lines)
print("Результат:", str(res_lst))
```

Напишите функцию, которая принимает в качестве аргумента первый узел односвязного
списка, обращает порядок элементов списка (с разрушением исходного) и возвращает в
качестве результата первый узел полученного списка. Разработайте два варианта решения
этой задачи: 1) итеративный и 2) рекурсивный.

```python
def reverse_iterative(head: Node) -> Node:
    prev = None
    cur = head
    while cur:
        nxt = cur.next
        cur.next = prev
        prev = cur
        cur = nxt
    return prev

def reverse_recursive(head: Node) -> Node:
    if head is None or head.next is None:
        return head
    rest = reverse_recursive(head.next)
    head.next.next = head
    head.next = None
    return rest

def list_from_array(arr):
    head = Node(arr[0])
    cur = head
    for x in arr[1:]:
        cur.next = Node(x)
        cur = cur.next
    return head

def print_list(head):
    res = []
    cur = head
    while cur:
        res.append(cur.data)
        cur = cur.next
    return res

head = list_from_array([1,2,3,4])
rev_iter = reverse_iterative(head)
print("Iterative reverse:", print_list(rev_iter))

head = list_from_array([1,2,3,4])
rev_rec = reverse_recursive(head)
print("Recursive reverse:", print_list(rev_rec))
```

```python
class Stack:
    def __init__(self):
        self.lst = UnorderedList()
    def push(self, item):
        self.lst.add(item)
    def pop(self):
        return self.lst.pop(0)
    def peek(self):
        return self.lst.head.data if self.lst.head else None
    def is_empty(self):
        return self.lst.is_empty()

class Queue:
    def __init__(self):
        self.lst = UnorderedList()
    def enqueue(self, item):
        self.lst.append(item)
    def dequeue(self):
        return self.lst.pop(0)
    def is_empty(self):
        return self.lst.is_empty()

class Deque:
    def __init__(self):
        self.lst = UnorderedList()
    def add_front(self, item):
        self.lst.add(item)
    def add_rear(self, item):
        self.lst.append(item)
    def remove_front(self):
        return self.lst.pop(0)
    def remove_rear(self):
        return self.lst.pop()
    def is_empty(self):
        return self.lst.is_empty()

s = Stack(); [s.push(x) for x in [1,2,3]]
print("Stack pop:", s.pop())

q = Queue(); [q.enqueue(x) for x in [1,2,3]]
print("Queue dequeue:", q.dequeue())

d = Deque(); [d.add_front(1), d.add_rear(2), d.add_front(3)]
print("Deque remove_front:", d.remove_front())
print("Deque remove_rear:", d.remove_rear())
```

Реализуйте класс DoubleList для построения двухсвязных списков, в которых каждый узел
содержит ссылку на узел, предшествующий ему в списке, и ссылку на следующий узел.
Класс должен иметь следующие методы: проверка списка на пустоту, поиск элемента,
размер списка, вставка в начало, вставка в конец, удаление из начала, удаление из конца,
вставка перед указанным узлом, вставка после указанного узла, удаление указанного узла.
Проведите тестирование методов реализованного класса.

```python
class DNode:
    def __init__(self, data: Any):
        self.data = data
        self.prev = None
        self.next = None

class DoubleList:
    def __init__(self):
        self.head = None
        self.tail = None
        self._size = 0

    def is_empty(self):
        return self._size == 0

    def size(self):
        return self._size

    def add_first(self, item):
        n = DNode(item)
        n.next = self.head
        if self.head:
            self.head.prev = n
        self.head = n
        if self.tail is None:
            self.tail = n
        self._size += 1

    def add_last(self, item):
        n = DNode(item)
        n.prev = self.tail
        if self.tail:
            self.tail.next = n
        self.tail = n
        if self.head is None:
            self.head = n
        self._size += 1

    def remove_first(self):
        val = self.head.data
        self.head = self.head.next
        if self.head:
            self.head.prev = None
        else:
            self.tail = None
        self._size -= 1
        return val

    def remove_last(self):
        val = self.tail.data
        self.tail = self.tail.prev
        if self.tail:
            self.tail.next = None
        else:
            self.head = None
        self._size -= 1
        return val

    def find(self, item):
        cur = self.head
        while cur:
            if cur.data == item:
                return True
            cur = cur.next
        return False

    def insert_before(self, target, item):
        cur = self.head
        while cur:
            if cur.data == target:
                n = DNode(item)
                n.prev = cur.prev
                n.next = cur
                if cur.prev:
                    cur.prev.next = n
                else:
                    self.head = n
                cur.prev = n
                self._size += 1
                return
            cur = cur.next

    def insert_after(self, target, item):
        cur = self.head
        while cur:
            if cur.data == target:
                n = DNode(item)
                n.next = cur.next
                n.prev = cur
                if cur.next:
                    cur.next.prev = n
                else:
                    self.tail = n
                cur.next = n
                self._size += 1
                return
            cur = cur.next

    def remove(self, item):
        cur = self.head
        while cur:
            if cur.data == item:
                if cur.prev:
                    cur.prev.next = cur.next
                else:
                    self.head = cur.next
                if cur.next:
                    cur.next.prev = cur.prev
                else:
                    self.tail = cur.prev
                self._size -= 1
                return
            cur = cur.next

    def __str__(self):
        res = []
        cur = self.head
        while cur:
            res.append(cur.data)
            cur = cur.next
        return str(res)

dl = DoubleList()
dl.add_first(1); dl.add_last(2); dl.add_last(3)
print("DoubleList:", str(dl))
dl.insert_before(2, 99)
print("After insert_before 2:", str(dl))
dl.insert_after(2, 88)
print("After insert_after 2:", str(dl))
print("remove_first:", dl.remove_first(), str(dl))
print("remove_last:", dl.remove_last(), str(dl))
dl.remove(2)
print("After remove 2:", str(dl))
```

быть сбалансированы, чтобы адекватно описывать веб-документ. Вот пример очень
простого веб-документа:
<html>
 <head>
 <title>
 Example
 </title>
 </head>
 <body>
 <h1>Hello, world</h1>
 </body>
</html>
Напишите программу, которая, используя структуру данных стек, проверит HTML
документ на соответствие открывающих и закрывающих тегов.
За основу решения можно взять алгоритм проверки скобочной последовательности из
лекционных материалов.

```python

def check_html_balance(html_lines):
    stack = []
    for line in html_lines:
        line = line.strip()
        while '<' in line:
            start = line.find('<')
            end = line.find('>', start)
            if end == -1:
                return False
            tag_content = line[start+1:end].strip()
            if not tag_content.startswith('/'):
                tag = tag_content
                stack.append(tag)
            else: 
                tag = tag_content[1:]
                if not stack or stack[-1] != tag:
                    return False
                stack.pop()
            line = line[end+1:] 
    return len(stack) == 0

html_doc = [
    "<html>",
    "<head>",
    "<title>Example</title>",
    "</head>",
    "<body>",
    "<h1>Hello, world</h1>",
    "</body>",
    "</html>"
]

html_doc_unbalanced = [
    "<html>", "<body>", "<h1>Hello</h1>", "</html>"
]

print("HTML сбалансирован:", check_html_balance(html_doc))         # True
print("HTML несбалансирован:", check_html_balance(html_doc_unbalanced))  # False
```

Стековый калькулятор - это устройство (реальное или виртуальное), которое вычисляет
значения арифметических выражений, записанных в постфиксной форме (также она носит
название Обратная польская запись — Википедия (wikipedia.org)). Стековым он называется
так как действует по следующим правилам: если нам встречается число - мы должны
поместить его в стек; если знак операции - мы должны достать из стека столько чисел,
сколько необходимо для выполнения данной операции и выполнить ее.
Именно так и работает т.н. математический сопроцессор (например, в INtel8086+AMD64)-
часть ЦП, отвечающая за операции над числами с плавающей запятой.
Реализуйте стековый калькулятор. Напишите программу, которая читает выражение в
обратной польской нотации и выводит получившееся значение или пишет, что выражение
составлено не корректно (если оно некорректно).

```python
def rpn_calculator(expr_tokens):
    stack = []
    for token in expr_tokens:
        if token.isdigit() or (token.startswith('-') and token[1:].isdigit()):
            stack.append(int(token))
        else:
            if len(stack) < 2:
                return "Некорректное выражение"
            b = stack.pop()
            a = stack.pop()
            if token == '+':
                stack.append(a + b)
            elif token == '-':
                stack.append(a - b)
            elif token == '*':
                stack.append(a * b)
            elif token == '/':
                stack.append(a / b)
            else:
                return "Неизвестная операция"
    if len(stack) != 1:
        return "Некорректное выражение"
    return stack[0]

expr1 = ["3", "4", "+", "2", "*"]  # (3+4)*2 = 14
expr2 = ["5", "1", "2", "+", "4", "*", "+", "3", "-"]  # 5+((1+2)*4)-3 = 14
expr_invalid = ["3", "+"]

print("Результат expr1:", rpn_calculator(expr1))
print("Результат expr2:", rpn_calculator(expr2))
print("Результат некорректного выражения:", rpn_calculator(expr_invalid))
```

```python
class Queue:
    def __init__(self):
        self.items = []
    def enqueue(self, item):
        self.items.append(item)
    def dequeue(self):
        if not self.is_empty():
            return self.items.pop(0)
    def is_empty(self):
        return len(self.items) == 0
    def size(self):
        return len(self.items)

import random

def car_wash_simulation(n_cars, wash_time_mean=5):
    queue = Queue()
    for i in range(1, n_cars+1):
        queue.enqueue(f"Car{i}")
    times = []
    while not queue.is_empty():
        car = queue.dequeue()
        time_spent = random.randint(3, 7)  # случайное время мытья 3-7 минут
        times.append((car, time_spent))
    return times

random.seed(42)
results = car_wash_simulation(5)
for car, time_spent in results:
    print(f"{car} помыта за {time_spent} минут")
```
